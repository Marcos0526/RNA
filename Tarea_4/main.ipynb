```python

# Paso 2: Entrenar clasificador con tus fotos (Capas congeladas)
train_face, val_face = get_face_recognition_generators('dataset/faces/')
face_model = build_face_classifier('models/base_celeba_model.keras', freeze_features=True)
face_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
face_model.fit(train_face, validation_data=val_face, epochs=10)

# Paso 3: Fine-Tuning
face_model = prepare_for_fine_tuning(face_model, unfreeze_from_layer=-3, learning_rate=1e-5)
face_model.fit(train_face, validation_data=val_face, epochs=5)
```
